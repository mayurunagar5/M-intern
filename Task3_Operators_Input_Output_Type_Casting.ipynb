## Arithmetic Operators
Explanation and example.


```python
a=10
b=3
print(a+b,a-b,a*b,a/b,a//b,a%b,a**b)
```

## Assignment Operators
Explanation and example.


```python
x=10
x+=5
print(x)
x*=2
print(x)
```

## Comparison Operators
Explanation and example.


```python
a=10
b=20
print(a==b,a!=b,a<b,a>=b)
```

## Logical Operators
Explanation and example.


```python
age=20
citizen=True
print(age>=18 and citizen)
print(age<18 or citizen)
print(not citizen)
```

## Bitwise Operators
Explanation and example.


```python
a=5
b=3
print(a&b,a|b,a^b,~a,a<<1,a>>1)
```

## Membership Operators
Explanation and example.


```python
fruits=['Apple','Banana']
print('Apple' in fruits)
print('Orange' not in fruits)
```

## Identity Operators
Explanation and example.


```python
a=[1,2]
b=a
c=[1,2]
print(a is b)
print(a is c)
```

## Input and Output

## Type Casting
Explanation and example.


```python
age='20'
print(int(age)+5)
print(float('2.5'))
print(str(100))
print(bool(1), bool(0))
```

## Implicit vs Explicit
Explanation and example.


```python
a=10
b=2.5
print(a+b)
print(int('50'))
```
