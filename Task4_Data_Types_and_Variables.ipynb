```python
x=10
print(type(x))
x='Python'
print(type(x))
```

## Integer
Explanation with example.


```python
num=100
print(num)
print(type(num))
```

## Float
Explanation with example.


```python
price=99.99
print(price)
print(type(price))
```

## Complex
Explanation with example.


```python
z=2+3j
print(z)
print(type(z))
```

## Boolean
Explanation with example.


```python
is_student=True
print(is_student)
print(type(is_student))
```

## String
Explanation with example.


```python
text='Hello Python'
print(text)
print(text[0])
print(text.upper())
```

## List
Explanation with example.


```python
fruits=['Apple','Banana','Mango']
fruits.append('Orange')
print(fruits)
print(fruits[1])
```

## Tuple
Explanation with example.


```python
nums=(10,20,30)
print(nums)
print(nums[0])
```

## Set
Explanation with example.


```python
colors={'Red','Blue','Green','Red'}
colors.add('Yellow')
print(colors)
```

## Dictionary
Explanation with example.


```python
student={'name':'Mayur','age':20}
print(student)
print(student['name'])
```

## None Type
Explanation with example.


```python
value=None
print(value)
print(type(value))
```

## Naming Conventions
Explanation with example.


```python
student_name='Mayur'
TOTAL_MARKS=100
print(student_name)
print(TOTAL_MARKS)
```

## Multiple Assignment
Explanation with example.


```python
a,b,c=10,20,30
print(a,b,c)
x=y=z=5
print(x,y,z)
a,b=b,a
print(a,b)
```

## Type Checking
Explanation with example.


```python
print(type(10))
print(type(10.5))
print(type('Hello'))
print(type([1,2]))
print(type((1,2)))
print(type({1,2}))
print(type({'a':1}))
```

## Practice Program
Explanation with example.


```python
name='Mayur'
age=20
skills=['Python','Git']
profile={'name':name,'age':age,'skills':skills}
print(profile)
for s in skills:
    print(s)
```
