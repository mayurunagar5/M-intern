## 1. Indexing
Indexing accesses specific elements.
- 1D: arr[0]
- 2D: arr[row, col]
- 3D: arr[layer, row, col]
- Negative indexing: arr[-1]
- Fancy indexing and Boolean indexing.

```python
import numpy as np
arr=np.array([[10,20,30],[40,50,60],[70,80,90]])
print(arr[0,1])
print(arr[:,1])
print(arr[[0,2]])
```

## 2. Slicing
Syntax: arr[start:stop:step]
For 2D: arr[rows, cols]

```python
print(arr[0:2,1:3])
print(arr[::-1])
```

## 3. Broadcasting
Broadcasting automatically expands compatible shapes during arithmetic.

```python
a=np.array([[1,2,3],[4,5,6]])
b=np.array([10,20,30])
print(a+b)
```

## 4. Vectorization
Operate on entire arrays without Python loops.

```python
x=np.array([1,2,3,4])
print(x*5)
print(x**2)
```

## 5. Universal Functions (ufuncs)

```python
print(np.sqrt(x))
print(np.square(x))
print(np.mean(x))
print(np.max(x))
```

## 6. Boolean Masking
Filter arrays using conditions.

```python
marks=np.array([45,67,82,91,58])
print(marks[marks>=60])
```

## 7. Performance Comparison

```python
from timeit import timeit
py=timeit("[i*2 for i in range(10000)]",number=100)

np_time=timeit("arr*2",setup="import numpy as np; arr=np.arange(10000)",number=100)
print("Python:",py)
print("NumPy :",np_time)
```
