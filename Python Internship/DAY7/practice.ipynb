```python
#iterator

# We create an iterator using iter() and get values using next()

numbers = [10, 20, 30, 60]
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
```

```python
# Generator prac.
def roll_number():
    yield 101
    yield 102
    yield 103
    yield 104


students= roll_number()
print(next(students))
print(next(students))

print(next(students))
print(next(students))
```

```python
def square(limit):
    for number in range(1, limit+1):
        yield number**2

result = square(5)
print(next(result))
print(next(result))

print(next(result))
print(next(result))
print(next(result))
```

```python
def even_numbers(limit):

    for number in range(2, limit + 1, 2):
        yield number

for value in even_numbers(10):
    print(value)
```

```python
def names():
    names = ["Mayur", "Vatsal", "Vinit"]

    for name in names:
        yield name

result = names()

print(next(result))
print(next(result))
print(next(result))
```

```python
orders = [
    "Order 1",
    "Order 2",
    "Order 3",
    "Order 4",
    "Order 5"
]

def get_order():
    for o_id in range(1,6):
        print(f"Loading order : {o_id}")
        yield o_id

orders = get_order()

for order in orders:
    print(f"Processing :", order)
```

```python
def load_images():

    images = [
        "cat1.jpg",
        "cat2.jpg",
        "cat3.jpg",
        "cat4.jpg",
        "cat5.jpg"
    ]

    for image in images:
        print(f"Loading {image}")

        yield image


for image in load_images():
    print("Training on", image)
```
