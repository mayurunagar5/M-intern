```python
# iterable

fruits = ["banana", "apple", "orange"]
for fruit in fruits:
    print(fruit)
```

```python
#iterator

# We create an iterator using iter() and get values using next()

numbers = [10, 20, 30, 60]
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
```

```python
# Generator prac.
def roll_number():
    yield 101
    yield 102
    yield 103
    yield 104


students= roll_number()
print(next(students))
print(next(students))

print(next(students))
print(next(students))
```

```python
def square(limit):
    for number in range(1, limit+1):
        yield number**2

result = square(5)
print(next(result))
print(next(result))

print(next(result))
print(next(result))
print(next(result))
```

```python
def even_numbers(limit):

    for number in range(2, limit + 1, 2):
        yield number

for value in even_numbers(10):
    print(value)
```
