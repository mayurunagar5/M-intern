## 1. Iterable
An iterable is any object that can be looped using `for`.

Examples: list, tuple, string, set, dictionary.

```python
fruits=['Apple','Banana','Mango']
for fruit in fruits:
    print(fruit)
```

## 2. Iterator
An iterator is created using `iter()` and returns one item at a time using `next()`.

```python
numbers=[10,20,30]
it=iter(numbers)
print(next(it))
print(next(it))
print(next(it))
```

## Iterable vs Iterator
| Iterable | Iterator |
|---|---|
| Can be looped | Gives one value at a time |
| Used with `for` | Used with `next()` |
| list, tuple, str | Created by `iter()` |

## 3. Generator
Generators use `yield` instead of `return`. They generate values one by one and save memory.

```python
def even(limit):
    for n in range(2,limit+1,2):
        yield n

for value in even(10):
    print(value)
```

## return vs yield
- `return` ends the function.
- `yield` pauses the function and resumes from the same place.

## 4. Decorator
A decorator adds functionality to another function without changing its original code.

```python
def decorator(func):
    def wrapper():
        print('Before Function')
        func()
        print('After Function')
    return wrapper

@decorator
def greet():
    print('Hello Mayur!')

greet()
```

## Real-world Uses
- Decorators: Authentication, logging, timing, Flask/FastAPI routes.
- Generators: Large datasets, file reading, ML pipelines.
- Iterators: Custom iteration and `for` loops.
