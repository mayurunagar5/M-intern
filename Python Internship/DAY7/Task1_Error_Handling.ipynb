## Error Handling & Exception Handling

```python
try:
    print(10/0)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    print(10/2)
except ZeroDivisionError:
    print("Error")
else:
    print("Success!")
```

```python
try:
    value=int("5")
finally:
    print("Program Finished")
```

```python
class AgeError(Exception):
    pass

try:
    age=10
    if age>18:
        raise AgeError("Eligible")
    else:
        raise AgeError("Not Eligible")
    
except AgeError as e:
    print(e)   
```

```python
class AgeError(Exception):
    pass

try:
    age=16
    if age<18:
        raise AgeError("Not Eligible")
except AgeError as e:
    print(e)
```

```python
a = int(input("Enter a number: "))
b = int(input("Enter a number: "))

try:
    result = a/b
except ZeroDivisionError:   
    print("Cannot divide by zero.") 
```

```python

try:
    Age = int(input("Enter a number: "))

    print(Age)

except ValueError:
    print("Please enter valid number")
```

```python
numbers = [10, 20, 50,70, 90]

try:
    index = int(input("Enter your index:"))
    print(numbers[index])
except IndexError:
    print("Invalid index")
```

```python
student = {
    "name": "Mayur",
    "age": 20,
    "city": "Surat"
}

try:
    key = input("Enter Key: ")
    print(student[key])

except KeyError:
    print("Key does not exist.")
```

```python
try:
    number = int(input("Enter Number: "))
    print(100 / number)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```
