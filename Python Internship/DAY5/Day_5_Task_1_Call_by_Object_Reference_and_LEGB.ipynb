## Call-by-Object-Reference & Scope Rules (LEGB)

### Immutable Objects

```python
def increase(number):
    number += 10
    print('Inside:', number)

value = 20
increase(value)
print('Outside:', value)
```

### Mutable Objects

```python
def add_item(items):
    items.append('Python')

languages=['Java']
add_item(languages)
print(languages)
```

### Global Scope

```python
name='Mayur'

def display():
    print(name)

display()
```

### global Keyword

```python
count=10

def update():
    global count
    count=20

update()
print(count)
```

### Enclosing Scope & nonlocal

```python
def outer():
    count=10
    def inner():
        nonlocal count
        count+=5
    inner()
    print(count)

outer()
```

### Built-in Scope

```python
numbers=[10,20,30]
print(len(numbers))
print(max(numbers))
print(sum(numbers))
```

## Summary
- Call-by-Object-Reference
- LEGB Rule: Local → Enclosing → Global → Built-in
- Use `global` to modify global variables.
- Use `nonlocal` to modify enclosing variables.
