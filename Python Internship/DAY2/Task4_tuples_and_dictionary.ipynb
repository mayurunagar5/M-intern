Tuple Operations Cheat Sheet

| Operation | Description | Example | Output |
|-----------|-------------|---------|--------|
| Create Tuple | Create a tuple | `(1,2,3)` | `(1,2,3)` |
| Single Element Tuple | Use comma | `(5,)` | `(5,)` |
| Indexing | Access by index | `t[1]` | `2` |
| Negative Indexing | Access from the end | `t[-1]` | `3` |
| Slicing | Extract a portion | `t[1:3]` | `(2,3)` |
| Concatenation | Join two tuples | `(1,2)+(3,4)` | `(1,2,3,4)` |
| Repetition | Repeat tuple | `(1,2)*2` | `(1,2,1,2)` |
| Membership | Check if an item exists | `2 in t` | `True` |
| Iteration | Loop through tuple | `for i in t` | Prints each item |
| Comparison | Compare tuples | `(1,2)==(1,2)` | `True` |
| Nested Tuple | Tuple inside tuple | `((1,2),(3,4))` | Nested tuple |
| Tuple Unpacking | Assign tuple values to variables | `a,b=(1,2)` | `a=1, b=2` |

#  Tuple Methods Cheat Sheet

> Tuples are **immutable**, so they have only **two built-in methods**.

| Method | Description | Example | Output |
|---------|-------------|---------|--------|
| `count()` | Counts the occurrences of a value | `(1,2,1).count(1)` | `2` |
| `index()` | Returns the first index of a value | `(10,20,30).index(20)` | `1` |

---


````python
#Count Example
grades = ("A", "B", "A", "C", "B", "A", "D", "A", "B", "C")

grade_a= grades.count("A")
grade_b= grades.count("B")
grade_c= grades.count("c")
grade_d= grades.count("D")

print("A: ", grade_a)
print("B: ", grade_b)
print("C: ", grade_c)
print("D: ", grade_d)
````

````python
# Index Example:

String = ("mayur", "vinit", "vatsal", "parth")

text = "parth"
result = String.index(text)
print(result)
````

> **Interview Tip:**  
> **List = Mutable** (can be modified)  
> **Tuple = Immutable** (cannot be modified)

# 🐍 Python Dictionary

A **Dictionary** is a built-in Python data type that stores data in **key-value pairs**.

## Characteristics

- ✅ Stores data as **key : value**
- ✅ Mutable (can be modified)
- ✅ Ordered (Python 3.7+)
- ✅ Keys must be unique
- ✅ Values can be duplicated
- ✅ Can store different data types
- ✅ Uses curly braces `{}`

````python
# Example:

student = {
    "name": "Mayur",
    "age": 20,
    "city": "Surat"
}

print(student)
````

# 📌 Dictionary Operations Cheat Sheet

| Operation | Description | Example | Output |
|-----------|-------------|---------|--------|
| Create Dictionary | Create a dictionary | `{"name":"Mayur","age":20}` | `{'name':'Mayur','age':20}` |
| Access Value | Access value by key | `student["name"]` | `Mayur` |
| Safe Access | Access value safely | `student.get("age")` | `20` |
| Add Item | Add new key-value pair | `student["city"]="Surat"` | Added |
| Update Item | Update an existing value | `student["age"]=21` | Updated |
| Delete Item | Delete using `del` | `del student["age"]` | Removed |
| Membership | Check if key exists | `"name" in student` | `True` |
| Length | Number of key-value pairs | `len(student)` | `3` |
| Iteration | Loop through dictionary | `for key in student:` | Prints keys |
| Nested Dictionary | Dictionary inside another | `students["s1"]["name"]` | `Mayur` |
| Dictionary Comprehension | Create dictionary with loop | `{x:x*x for x in range(5)}` | `{0:0,1:1...}` |

---

````python
# all operation with example:

# 1. Create Dictionary

details ={"name":"Mayur", "age":20, "city":"surat"}
print(details)
````

````python
# 2. Access Value (Using Key)
students ={"name":"Mayur", "age":20, "city":"surat"}

print(students["name"])
````

````python
# 3. Access Value using get()

students ={"name":"Mayur", "age":20, "city":"surat"}
print(students.get("age"))
print(students.get("enroll", "not found"))

````

````python
# 4. Add a New Key-Value Pair

students ={"name":"Mayur", "age":20, "city":"surat"}

students["fruit"]="mango"
print(students)
````

````python
# 5. Update an Existing Value

students ={"name":"Mayur", "age":20, "city":"surat"}
students["age"]=21
print(students)

````

````python
# 6. Delete a Key-Value Pair (del)

students ={"name":"Mayur", "age":20, "city":"surat"}
del students["city"]
print(students)
````

````python
# 7. Membership Operator (in, not in)

students ={"name":"Mayur", "age":20, "city":"surat"}

print("name" in "het")
print("surat" not in "city")
````

````python
# 8. Iterate Through Keys
students ={"name":"Mayur", "age":20, "city":"surat"}

for key in students.values():
    print(key)
````

````python
# 9. Nested Dictionary

students = {
    "student1": {
        "name": "Mayur",
        "age": 20
    },
    "student2": {
        "name": "Rahul",
        "age": 21
    }
}

print(students["student1"]["name"])
````

# 📚 Dictionary Methods Cheat Sheet

| Method | Description | Syntax |
|---------|-------------|--------|
| `clear()` | Remove all items | `dict.clear()` |
| `copy()` | Create a shallow copy | `dict.copy()` |
| `fromkeys()` | Create dictionary from keys | `dict.fromkeys(keys, value)` |
| `get()` | Return value of a key | `dict.get(key)` |
| `items()` | Return all key-value pairs | `dict.items()` |
| `keys()` | Return all keys | `dict.keys()` |
| `values()` | Return all values | `dict.values()` |
| `pop()` | Remove a key and return its value | `dict.pop(key)` |
| `popitem()` | Remove the last inserted key-value pair | `dict.popitem()` |
| `setdefault()` | Return value if key exists; otherwise insert default value | `dict.setdefault(key, value)` |
| `update()` | Update dictionary with another dictionary | `dict.update({...})` |

---
