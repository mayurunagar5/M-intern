# 1. String (`str`)

A **String** is an immutable sequence of characters enclosed in single (`' '`) or double (`" "`) quotes.

```python
text = "Python Programming"
```

## String Operations

```python
print(text[0])          # Indexing
print(text[0:6])        # Slicing
print(len(text))        # Length
print("Python" in text) # Membership
```

# 🐍 Python String Operations

String operations are used to manipulate and work with strings in Python.

---

# 1. Indexing

Access a single character using its index.

### Example 1

```python
text = "Python"

print(text[0])
```

**Output**

```
P
```

### Example 2

```python
text = "Programming"

print(text[-1])
```

**Output**

```
g
```

---

# 2. Slicing

Extract a portion of a string.

### Example 1

```python
text = "Python Programming"

print(text[0:6])
```

**Output**

```
Python
```

### Example 2

```python
text = "Python Programming"

print(text[7:])
```

**Output**

```

```

---

# 3. Concatenation (`+`)

Join two or more strings.

### Example 1

```python
first = "Hello"
second = "World"

print(first + " " + second)
```

**Output**

```
Hello World
```

### Example 2

```python
a = "Python"
b = "3"

print(a + " " + b)
```

**Output**

```
Python 3
```

---

# 4. Repetition (`*`)

Repeat a string multiple times.

### Example 1

```python
text = "Hi "

print(text * 3)
```

**Output**

```

```

### Example 2

```python

```

```

```

Insert variables into a string.

# 🐍 Python String Methods Cheat Sheet

| Method | Description | Example | Output |
|--------|-------------|---------|--------|
| `upper()` | Converts all characters to uppercase | `"python".upper()` | `PYTHON` |
| `lower()` | Converts all characters to lowercase | `"PYTHON".lower()` | `python` |
| `title()` | Capitalizes the first letter of every word | `"python programming".title()` | `Python Programming` |
| `capitalize()` | Capitalizes only the first letter of the string | `"python programming".capitalize()` | `Python programming` |
| `swapcase()` | Swaps uppercase to lowercase and vice versa | `"PyThOn".swapcase()` | `pYtHoN` |
| `strip()` | Removes whitespace from both ends | `" Python ".strip()` | `Python` |
| `lstrip()` | Removes whitespace from the left side | `" Python".lstrip()` | `Python` |
| `rstrip()` | Removes whitespace from the right side | `"Python ".rstrip()` | `Python` |
| `replace(old, new)` | Replaces a substring | `"I love Python".replace("Python","Java")` | `I love Java` |
| `split()` | Splits a string into a list | `"Python Java C++".split()` | `['Python', 'Java', 'C++']` |
| `join(iterable)` | Joins iterable elements into a string | `"-".join(["Python","Java"])` | `Python-Java` |
| `find()` | Returns the first index of a substring (`-1` if not found) | `"Python".find("th")` | `2` |
| `index()` | Returns the first index of a substring (raises error if not found) | `"Python".index("th")` | `2` |
| `count()` | Counts occurrences of a substring | `"programming".count("m")` | `2` |
| `startswith()` | Checks if the string starts with a prefix | `"Python".startswith("Py")` | `True` |
| `endswith()` | Checks if the string ends with a suffix | `"Python".endswith("on")` | `True` |
| `isalpha()` | Returns `True` if all characters are letters | `"Python".isalpha()` | `True` |
| `isdigit()` | Returns `True` if all characters are digits | `"12345".isdigit()` | `True` |
| `isalnum()` | Returns `True` if all characters are letters or digits | `"Python123".isalnum()` | `True` |
| `isspace()` | Returns `True` if all characters are whitespace | `"   ".isspace()` | `True` |
| `islower()` | Returns `True` if all letters are lowercase | `"python".islower()` | `True` |
| `isupper()` | Returns `True` if all letters are uppercase | `"PYTHON".isupper()` | `True` |
| `center(width)` | Centers the string within the specified width | `"Python".center(10)` | `'  Python  '` |
| `ljust(width)` | Left-aligns the string | `"Python".ljust(10)` | `'Python    '` |
| `rjust(width)` | Right-aligns the string | `"Python".rjust(10)` | `'    Python'` |
| `zfill(width)` | Pads the string with leading zeros | `"42".zfill(5)` | `00042` |

````python
text = "Python programming"

result = text.upper()

text = "PYTHON"
result = text.lower()
print(result)

# title()
text = "python programming"
result = text.title()
print(result)

# title()
text = "python programming"
result = text.capitalize()
print(result)

# swapcase()
text = "PyThOn"
result = text.swapcase()
print(result)

# strip()
text = "  Python  "
result = text.strip()
print(result)

# lstrip()
text = "   Python"
result = text.lstrip()
print(result)

# rstrip()
text = "Python   "
result = text.rstrip()
print(result)

# replace()
text = "I love Python"
result = text.replace("Python" , "Java")
print(result)

# split()
text = "Python Java C++"
result = text.split()
print(result)

# join()
languages = ["Python", "Java", "C++"]
result = "-".join(languages)
print(result)

# find()
text = "Python Programming"
result = text.find("Program")
print(result)

# index()
text = "Python Programming"
result = text.index("rogram")
print(result)

# count()
text = "programming"
result = text.count("m")
print(result)

# startswith()
text = "Python Programming"
result = text.startswith("Pthon")
print(result)

# endswith()
text = "Python Programming"
result = text.endswith("ing")
print(result)


# isalpha()
text = "Python"
result = text.isalpha()
print(result)

# isdigit()
text = "12345"
result = text.isdigit()
print(result)

# isalnum()
text = "Python123"
result = text.isdigit()
print(result)


# center()
text = "Python"
result = text.center(15)
print(result)


# ljust()
text = "Python"
result = text.ljust(15)
print(result)


# zfill()
text = "42"
result = text.zfill(5)
print(result)


# format()
result = "My name is {}".format("Mayur")
print(result)

# format()
result = "My name is {}".format("Mayur")
print(result)

# removeprefix() (Python 3.9+)
text = "unhappy"
result = text.removeprefix("un")
print(result)

# removesuffix() (Python 3.9+)
text = "filename.txt"
result = text.removesuffix(".txt")
print(result)


````
