# **Mutable Objects**

A mutable object can be changed after it is created.

You can:

Add elements


```python
fruits= ["Banana", "apple", "orange"]
fruits.append("Mango")
print(fruits)
```

**Remove Element**

```python
fruits = ["Banana","apple","orange"]
fruits.remove("Banana")
print(fruits)
```

**Update Element**

```python
fruits = ["Banana","apple","orange"]
fruits[1]= "grapes"
print(fruits)
```

# **Immutable Objects**

An immutable object cannot be changed after it is created.

If you appear to change it, Python actually creates a new object.

**String**

```python
name = "Mayur"
name = name + " Unagar"
print(name)
```

**Tupple Example**

```python
number= [10,20,30]
number[0] = 100
print(number)
```

**Memory Example Using id()**

The id() function shows an object's memory identity.

```python
numbers = [1, 2, 3]

print(id(numbers))

numbers.append(4)

print(id(numbers))
```
