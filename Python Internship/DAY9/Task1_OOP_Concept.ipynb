```python
# Bank system with Encapsulation : 

class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
    def deposite(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount<= self.__balance:
            self.__balance -= amount
    def balance_check(self):
        return self.__balance

bank = BankAccount("Mayur", 50000)

bank.deposite(5000)
bank.withdraw(10000)

print(bank.balance_check())

    
```

```python
# Super() : 

class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course
student = Student("Mayur", "AI - ML")

print(student.name)
print(student.course)
        
```

```python
class Employee:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Developer(Employee):
    def __init__(self, name, age, salary, course):
        super().__init__(name, age)
        self.salary = salary
        self.course = course


emp = Developer("Mayur",20, "ML", 500000)
print(emp.name)
print(emp.age)
print(emp.course)
print(emp.salary)
```

```python
# super() with calling parent method..

class Employee:
    def HR(self):
       print("hello")

class Developer(Employee):
    def HR(self):
        super().HR()
        print("leave")

emp = Developer()

emp.HR()
```

```python
# Multiple inheritance: 

class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Barking")


class Puppy(Dog):

    def play(self):
        print("Playing")


puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.play()
```

```python
# Hierarchical Inheritance : 

class Company:
    def sale(self):
        print("work started")

class employee(Company):
    def data(self):
        print("working")

class developer(Company):
    def task(self):
        print("Task completed")

emp = employee()
dev = developer()

emp.sale()
dev.sale()
```

```python
# Multi-level Inheritance :
class Camera:

    def photo(self):
        print("Take Photo")


class Phone:

    def call(self):
        print("Make Call")


class Smartphone(Camera, Phone):

    pass


mobile = Smartphone()

mobile.photo()
mobile.call()
```
