# 1. Class

## Definition

A **class** is a blueprint or template used to create objects. It defines the properties (attributes) and behaviors (methods) that every object created from it will have. A class itself does not store actual data; it only describes the structure of an object.

## Why do we use a Class?

- To organize code.
- To create multiple similar objects.
- To improve code reusability.
- To model real-world entities.

## Syntax

```python
class ClassName:
    pass
```

## Example

```python
class Car:

    def start(self):
        print("Car Started")

car = Car()
car.start()
```

### Output

```
Car Started
```

### Key Points

- A class is a blueprint.
- Objects are created from classes.
- One class can create many objects.

---

# 2. Object

## Definition

An **object** is an instance of a class. When you create an object, Python allocates memory for it. Each object has its own data while sharing the methods defined by the class.

## Why do we use Objects?

- To represent real-world entities.
- To store unique data.
- To call methods defined in the class.

## Example

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
```

### Output

```
<__main__.Student object at ...>
<__main__.Student object at ...>
```

Each object has a different memory location.

### Key Points

- Objects are created using a class.
- Every object has its own identity.
- Multiple objects can exist from one class.

# 3. Constructor (`__init__`)

## Definition

A constructor is a special method that automatically runs when an object is created. It initializes the object's data so that it is ready to use immediately.

## Why do we use a Constructor?

- To initialize object data.
- To avoid assigning values manually.
- To prepare an object when it is created.

## Syntax

```python
def __init__(self, ...):
```

## Example

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Mayur", 20)

print(student.name)
print(student.age)
```

### Output

```
Mayur
20
```

### Key Points

- Runs automatically.
- Used for initialization.
- Called only once during object creation.

# 4. self

## Definition

`self` refers to the **current object**. It allows an object to access its own attributes and methods. Python automatically passes `self` as the first argument when an instance method is called.

## Why do we use self?

- To access object attributes.
- To call other methods.
- To distinguish between different objects.

## Example

```python
class Student:

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student:", self.name)

student = Student("Mayur")
student.display()
```

### Output

```
Student: Mayur
```

### Key Points

- `self` represents the current object.
- It is passed automatically.
- It is required in every instance method.

---

# 5. Attributes

## Definition

Attributes are variables that belong to an object. They store information about the object, such as name, age, price, or color.

## Why do we use Attributes?

- To store object data.
- Every object can have different values.
- Attributes describe the state of an object.

## Example

```python
class Laptop:

    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

laptop = Laptop("Apple", 120000)

print(laptop.brand)
print(laptop.price)
```

### Output

```
Apple
120000
```

### Key Points

- Attributes store data.
- Created using `self`.
- Every object has its own attributes.

# 6. Methods (Instance Methods)

## Definition

Methods are functions defined inside a class. They describe the actions an object can perform. Instance methods always receive `self` as their first parameter.

## Why do we use Methods?

- To perform actions.
- To work with object data.
- To organize related functionality.

## Example

```python
class Car:

    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(self.brand, "Started")

car = Car("BMW")
car.start()
```

### Output

```
BMW Started
```

### Key Points

- Methods define object behavior.
- Methods can access attributes using `self`.
- Called using the object.

# 1. Encapsulation

## Definition

Encapsulation is the process of **combining data (attributes) and methods (functions)** into a single unit (class) while **restricting direct access** to the object's data. It protects the internal state of an object and allows controlled access through methods.

## Why do we use Encapsulation?

- Protects sensitive data.
- Prevents accidental modification.
- Makes code more secure.
- Improves maintainability.
- Controls how data is accessed.

## Key Concepts

- Public Members
- Protected Members (`_variable`)
- Private Members (`__variable`)
- Name Mangling
- Getter Methods
- Setter Methods

## Example

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount(5000)

account.deposit(2000)

print(account.get_balance())
```

### Output

```
7000
```

### Key Points

- Data is hidden.
- Access is controlled through methods.
- Uses private variables.

# 2. Inheritance

## Definition

Inheritance allows one class (Child Class) to acquire the properties and methods of another class (Parent Class). It promotes code reusability and establishes an "is-a" relationship between classes.

## Why do we use Inheritance?

- Reuse existing code.
- Reduce duplication.
- Extend existing functionality.
- Build hierarchical relationships.

## Types of Inheritance

- Single Inheritance
- Multilevel Inheritance
- Hierarchical Inheritance
- Multiple Inheritance
- Hybrid Inheritance

## Important Concepts

- Parent Class
- Child Class
- `super()`
- Method Overriding
- `isinstance()`
- `issubclass()`
- MRO (Method Resolution Order)

## Example

```python
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Barking")


dog = Dog()

dog.eat()
dog.bark()
```

### Output

```
Eating
Barking
```

### Key Points

- Child inherits Parent.
- Encourages code reuse.
- Supports method overriding.

---

# 3. Polymorphism

## Definition

Polymorphism means **"many forms."** It allows different objects to respond to the same method call in different ways. The same interface can perform different actions depending on the object.

## Why do we use Polymorphism?

- Makes code flexible.
- Reduces `if-else` conditions.
- Improves scalability.
- Supports reusable code.

## Types

- Method Overriding
- Duck Typing
- Built-in Polymorphism
- Operator Overloading

## Example

```python
class Dog:

    def sound(self):
        print("Dog Barks")


class Cat:

    def sound(self):
        print("Cat Meows")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

### Output

```
Dog Barks
Cat Meows
```

### Key Points

- Same method name.
- Different behavior.
- Improves flexibility.

---

# 4. Abstraction

## Definition

Abstraction means **hiding implementation details** and exposing only the necessary functionality. The user interacts with **what** an object does, without needing to know **how** it does it.

## Why do we use Abstraction?

- Hides complexity.
- Improves security.
- Makes programs easier to use.
- Separates interface from implementation.

## Important Concepts

- Abstract Class
- Abstract Method
- `ABC`
- `@abstractmethod`

## Example

```python
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Square(Shape):

    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


square = Square(4)

print(square.area())
```

### Output

```
16
```

### Key Points

- Hides implementation.
- Uses abstract classes.
- Forces child classes to implement required methods.

````python
# Multiple inheritance: 

class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Barking")


class Puppy(Dog):

    def play(self):
        print("Playing")


puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.play()
````

````python
# Hierarchical Inheritance : 

class Company:
    def sale(self):
        print("work started")

class employee(Company):
    def data(self):
        print("working")

class developer(Company):
    def task(self):
        print("Task completed")

emp = employee()
dev = developer()

emp.sale()
dev.sale()
````

````python
# Duck typing : 

class Email:

    def send(self):
        print("Sending Email")


class SMS:

    def send(self):
        print("Sending SMS")


class WhatsApp:

    def send(self):
        print("Sending WhatsApp Message")


def notify(service):
    service.send()


notify(Email())
notify(SMS())
notify(WhatsApp())
````
