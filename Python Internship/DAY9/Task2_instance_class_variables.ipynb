# Task 2: Instance Variables vs Class Variables

## Introduction
In Python, variables inside a class are divided into two types:

- Instance Variables
- Class Variables

Understanding the difference between them is essential because they differ in ownership, memory allocation, and usage.

---
## Instance Variables

### Definition

An instance variable belongs to an individual object. Each object has its own copy of the variable, so changing one object's value does not affect any other object.

### Syntax

```python
class Student:

    def __init__(self, name):
        self.name = name
```

### Example

```python
class Student:

    def __init__(self, name):
        self.name = name


student1 = Student("Mayur")
student2 = Student("Rahul")

print(student1.name)
print(student2.name)
```

### Output

```text
Mayur
Rahul
```

### Key Points

- Belongs to an object.
- Created using `self`.
- Every object has its own copy.

# `__str__()` vs `__repr__()`

| Feature | `__str__()` | `__repr__()` |
|---------|-------------|--------------|
| Purpose | Human-readable representation | Developer-friendly representation |
| Audience | End Users | Developers |
| Called By | `print(object)`, `str(object)` | `repr(object)`, Python Interpreter |
| Goal | Display information clearly | Show detailed object representation |
| Output Style | Simple and readable | More detailed and unambiguous |
| Main Use | User interface, reports, messages | Debugging and development |
| Return Type | Must return a string | Must return a string |
| If Missing | Python uses `__repr__()` as a fallback | Python shows the default object representation if `__repr__()` is also not defined |

---

# Quick Revision

- `__str__()` → **Readable for Users**
- `__repr__()` → **Detailed for Developers**
- `print(obj)` → Calls `__str__()`
- `repr(obj)` → Calls `__repr__()`
- If `__str__()` is missing, `print(obj)` usually falls back to `__repr__()`

````python
#Difference between __str__ and __repr__ :


class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age})"

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


student = Student("Mayur", 20)

print(student)
print(repr(student))
````

# `__len__()` - Key Points

- `__len__()` is a **Magic (Dunder) Method**.
- It defines the **length** of a custom object.
- Python automatically calls it when you use `len(object)`.
- It **must return a non-negative integer (`int`)**.
- If `__len__()` is not defined, `len(object)` raises a `TypeError`.
- It can return the length of a list, tuple, dictionary, string, or any value you define as the object's length.
- It makes your custom class behave like Python's built-in collections.
- It is an example of **built-in polymorphism** because the same `len()` function works with different object types.

---

## Flow

```text
len(object)
      │
      ▼
object.__len__()
      │
      ▼
Returns an Integer
```

---

## Common Mistakes

❌ Defining `__len__()` inside `__init__()`

❌ Returning a string

```python
return "5"
```

❌ Returning a negative number

```python
return -1
```

---


````python

class Library:

    def __init__(self):
        self.books = [
            "Python",
            "Java",
            "C++"
        ]

    def __len__(self):
        return len(self.books)


library = Library()

print(len(library))
````
