# 1. `sys` Module

The `sys` module provides access to the Python interpreter.

| Method / Attribute | Syntax | Returns | Purpose | Example | Common Use |
|---|---|---|---|---|---|
| `version` | `sys.version` | `str` | Python version | `print(sys.version)` | Version check |
| `platform` | `sys.platform` | `str` | Operating system | `print(sys.platform)` | Cross-platform code |
| `path` | `sys.path` | `list` | Import search paths | `print(sys.path)` | Debug imports |
| `argv` | `sys.argv` | `list[str]` | CLI arguments | `python app.py hi` | CLI tools |
| `exit()` | `sys.exit()` | None | Exit program | `sys.exit()` | Validation |
| `stdin` | `sys.stdin` | Stream | Standard input | `input()` | Keyboard |
| `stdout` | `sys.stdout.write()` | int | Standard output | `sys.stdout.write('Hi')` | Printing |
| `stderr` | `sys.stderr` | Stream | Error output | `print(err,file=sys.stderr)` | Logging |

# 2. `pathlib` Module

Modern replacement for many `os.path` operations.

| Method | Syntax | Returns | Purpose | Example |
|---|---|---|---|---|
| `Path()` | `Path('data')` | `Path` | Create path object | `Path('logs')` |
| `/` | `Path('data')/'train.csv'` | `Path` | Join paths | `Path('a')/'b'` |
| `exists()` | `path.exists()` | `bool` | Check existence | `path.exists()` |
| `is_file()` | `path.is_file()` | `bool` | Check file | `path.is_file()` |
| `is_dir()` | `path.is_dir()` | `bool` | Check directory | `path.is_dir()` |
| `mkdir()` | `path.mkdir()` | None | Create folder | `Path('logs').mkdir()` |
| `iterdir()` | `path.iterdir()` | Iterator | Iterate directory | `for p in Path('.').iterdir()` |
| `glob()` | `path.glob('*.csv')` | Iterator | Pattern matching | `glob('*.txt')` |
| `read_text()` | `path.read_text()` | str | Read file | `Path('a.txt').read_text()` |
| `write_text()` | `path.write_text()` | int | Write file | `Path('a.txt').write_text('Hi')` |

# Best Practices

- Prefer `pathlib` over `os.path`.
- Use `Counter` for frequencies.
- Use `deque` for queues.
- Use `lru_cache` for recursive optimization.
- Use `statistics` for simple descriptive statistics.
