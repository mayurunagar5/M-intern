Functional programming treats **functions as values**. Functions can be stored, passed, returned, and used to transform data.


## 1. First-Class Functions
Functions can be assigned to variables, passed to other functions, and returned from functions.

```python
def greet():
    return "Hello"

message = greet
print(message())
```

## 2. Higher-Order Functions
A higher-order function accepts another function as an argument or returns one.

```python
def square(n):
    return n*n

def calculate(func,value):
    return func(value)

print(calculate(square,5))
```

## 3. Lambda Functions
A lambda is a small anonymous function used for short operations.

```python
cube = lambda n: n**3
print(cube(3))
```

## 4. map()
Applies a function to every element of an iterable.

```python
marks=[60,70,80]
updated=list(map(lambda m:m+5,marks))
print(updated)
```

## 5. filter()
Returns only elements that satisfy a condition.

```python
numbers=[10,15,20,25,30]
even=list(filter(lambda n:n%2==0,numbers))
print(even)
```

## 6. reduce()
Combines all elements into one value.

```python
from functools import reduce
prices=[500,1200,800]
total=reduce(lambda a,b:a+b,prices)
print(total)
```

## 7. Closures
An inner function remembers variables from the outer function.

```python
def bonus(amount):
    def add(salary):
        return salary+amount
    return add

bonus5000=bonus(5000)
print(bonus5000(30000))
```

## 8. Decorators
Adds functionality to a function without changing its original code.

```python
def logger(func):
    def wrapper():
        print("Program Started")
        func()
        print("Program Ended")
    return wrapper

@logger
def display():
    print("Welcome")

display()
```

## Real-World Uses

- **Lambda:** Sorting, quick calculations.
- **map():** Increase salaries, convert temperatures.
- **filter():** Keep valid records, passed students.
- **reduce():** Total bill, total marks.
- **Closures:** Configuration, counters.
- **Decorators:** Authentication, logging, timing APIs.

## Quick Summary

```python
# 7. Decorators

def decorator(function):

    def wrapper(args):

        print("Program Started")

        function(args)

        print("Program Finished")

    return wrapper

@decorator
def welcome(args):
    print("Welcome Mayur")
    print(args)


# welcome = decorator(welcome)

welcome("vatsal")
```
