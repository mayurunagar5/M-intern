#  Python Slicing - Complete Guide

## What is Slicing?

**Slicing** is a technique used to extract a portion (subset) of a sequence in Python.

Slicing works on:

-  String
-  List
-  Tuple
-  Range

Slicing does **not** work directly on:

- Set
- Dictionary

---

# Syntax

```python
sequence[start : stop : step]
```

| Parameter | Description |
|-----------|-------------|
| start | Starting index (Included) |
| stop | Ending index (Excluded) |
| step | Number of elements to skip |

# Dictionary

```python
student = {
    "name":"Mayur",
    "age":20,
    "city":"Surat"
}
```

Invalid

```python
student[0:2]
```

Output

```text
TypeError
```

---

## Slice Dictionary Keys

```python
keys = list(student.keys())

print(keys[:2])
```

Output

```text
['name','age']
```

---

## Slice Dictionary Values

```python
values = list(student.values())

print(values[:2])
```

Output

```text
['Mayur',20]
```

---

## Slice Dictionary Items

```python
items = list(student.items())

print(items[:2])
```

Output

```text
[('name','Mayur'),('age',20)]
```

# Slicing Cheat Sheet

| Syntax | Meaning |
|---------|---------|
| `x[:]` | Copy the sequence |
| `x[::-1]` | Reverse the sequence |
| `x[:5]` | First 5 elements |
| `x[5:]` | Elements from index 5 to end |
| `x[2:8]` | Elements from index 2 to 7 |
| `x[::2]` | Every 2nd element |
| `x[1::2]` | Every 2nd element starting from index 1 |
| `x[-3:]` | Last 3 elements |
| `x[:-3]` | Everything except the last 3 elements |
| `x[1:-1]` | Remove first and last elements |
| `x[::-2]` | Reverse and skip every second element |

---

# Key Points

- Slicing works on **String, List, Tuple, and Range**.
- The **start index is included**.
- The **stop index is excluded**.
- Negative indexing starts from the end of the sequence.
- `[::-1]` is the most common way to reverse a sequence.
- Sets and Dictionaries do **not** support direct slicing because they are not index-based.
- You can slice dictionary **keys, values, or items** by first converting them into a list.

---

# Memory Trick

```
sequence[start : stop : step]
```

Remember:

- **Start → Included**
- **Stop → Excluded**
- **Step → Skip elements**
- **Negative Step → Reverse direction**
