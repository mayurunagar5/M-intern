#  Advanced Loops & Iteration Utilities

Advanced loop utilities help make Python code **cleaner, shorter, and more Pythonic**. They reduce the need for manual indexing and simplify iteration over multiple sequences.

---

# 1. `else` with `for` Loop

The `else` block executes **only if the `for` loop finishes normally**, meaning **no `break` statement is executed**.

### Syntax

```python
for item in iterable:
    # loop body
else:
    # executes if loop completes without break
```
### Example (No `break`)

````python
for number in range(1, 6):
    print(number)
else:
    print("Loop completed successfully")

````

```python
for number in range(1, 6):
    if number == 3:
        break
    print(number)
else:
    print("Loop completed successfully!")
```

**Output**

```text
1
2
```

# 2. `else` with `while` Loop

The `else` block in a `while` loop works exactly like a `for` loop.

It runs **only when the loop ends naturally**, not when `break` is used.

### Syntax

```python
while condition:
    # loop body
else:
    # executes if no break occurs
```

````python
count = 1
while count<=3:
    print(count)
    count+=1
else:
    print("Loop finished") 
````

# 3. `zip()` Function

The `zip()` function combines **two or more iterables** and iterates over them **in parallel**.

### Syntax

```python
zip(iterable1, iterable2, ...)
```

### Example

```python
students = ["Mayur", "Rahul", "Amit"]
marks = [90, 85, 88]

for student, mark in zip(students, marks):
    print(student, ":", mark)
```

**Output**

```text
Mayur : 90
Rahul : 85
Amit : 88
```

### Uses

- Combine two or more lists
- Pair names with marks
- Compare multiple lists
- Avoid manual indexing

---



````python
students = ["Mayur", "vatsal", "vinit"]
grades = ["A","b","c"]

for student, grade in zip(students, grades):
    print(f"{student}: {grade}")
````

# 4. `enumerate()` Function

The `enumerate()` function returns both the **index** and the **value** while iterating.

### Syntax

```python
enumerate(iterable, start=0)
```
### Example

```python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

**Output**

```text
0 Apple
1 Banana
2 Mango
```

````python
names = ["m","v","h", "u"]

for index, name in enumerate(names):
    print(index, name)
````

# 5. Using `zip()` and `enumerate()` Together

You can combine both functions to iterate over multiple lists **with automatic indexing**.

### Example

```python
students = ["Mayur", "Rahul", "Amit"]
marks = [90, 85, 88]

for index, (student, mark) in enumerate(zip(students, marks), start=1):
    print(index, student, mark)
```

**Output**

```text
1 Mayur 90
2 Rahul 85
3 Amit 88
```

# Quick Comparison

| Function | Purpose |
|----------|---------|
| `for...else` | Executes `else` only if the loop finishes without `break`. |
| `while...else` | Executes `else` only if the loop finishes without `break`. |
| `zip()` | Iterates over multiple iterables simultaneously. |
| `enumerate()` | Returns both index and value during iteration. |
| `enumerate(zip())` | Combines parallel iteration with automatic indexing. |

````python
# practice code

names = ["m", "v", "h"]
ages = [20,25,27]
cities= ["surat", "vadodara", "ahmedabad"]

for name, age, city in zip(names, ages, cities):
    print(f"{name} {age} {city}")
````

````python
number = [1,5,10,84,2,5,1]

for index, value in enumerate(number):
    if index % 2 ==0:
        number[index] = value * 2
    
print(number)
````

````python
product = ['m','h','p','u']
price = [10,20,30,40,50]
quantity = [2,6,7,8,9]

for pro, pri, qunty in zip(product, price, quantity):
    total = pri * qunty
    print(f"{pro}: {total}")
````
