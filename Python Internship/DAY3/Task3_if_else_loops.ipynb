#  Python Control Flow

## What is Control Flow?

Control Flow determines the **order in which Python executes statements**. By default, Python executes code from **top to bottom**, but using control flow statements we can make decisions, repeat tasks, or skip specific code.

### Types of Control Flow

- Decision Making (`if`, `elif`, `else`)
- Loops (`for`, `while`)
- Comprehensions (List, Dictionary, Set)

---

# 1. if Statement

Executes a block of code **only if the condition is True**.

### Syntax

```python
if condition:
    # code
```

### Example

```python
age = 20

if age >= 18:
    print("Eligible to Vote")
```

---

# 2. if...else Statement

Used when there are **two possible outcomes**.

### Syntax

```python
if condition:
    # True block
else:
    # False block
```



````python
#example:

age = 20
if age>= 18:
    print("You're eligible for voting")
else:
    print("not eligible")
````

 # 3. if...elif...else Statement

Used to check **multiple conditions**.


````python
marks = 80
if marks>=90:
    print("grade A")
elif marks>=80 and marks<90:
    print("grade B")
elif marks>=70 and marks<80:
    print("grade C")
else:
    print("Fail")
````

# 4. Nested if

An `if` statement inside another `if`.

```python

````python
age = 20
has_license = True

if age >= 18:
    if has_license:
        print("Can Drive")
````

````python
age = int(input("Enter yout age: "))
licence = input("You have driving licence yes or no: ")

if age>=18:
    if licence.lower() == "yes":
        print("You can drive")
    else:
        print("You need a driving licence.")
else:
    print("You are to young") 
````

# 5. for Loop

Used to iterate over a sequence such as a list, tuple, string, or range.

### Syntax

```python
for variable in iterable:
    # code
```

### Example

```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

---


````python
count = input("Enter yout name:")

for char in count:
    print(char)
````


# range()

`range()` generates a sequence of numbers.

Output:

````python
limit = int(input("Enter your limit:"))

for number in range(1, limit +1):
    print(number)
````

# 6. while Loop

Repeats a block of code **while the condition is True**.

### Syntax

```python
while condition:
    # code
```

````python
count = int(input("Enter the starting number: "))

while count <= 10:
    print(count)
    count += 1
    
````

# Loop Control Statements

## break

Stops the loop immediately.

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

---

````python
while True:
    password = input("Enter your password:")
    if password == "Python":
        print("Access granted")
        break;
    print("wrong password")
````

## continue

Skips the current iteration.

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

---

````python
for number in range(1, 11):

    if number == 5:
        continue

    print(number)
    
````

## pass

Acts as a placeholder. It does nothing.

```python
for i in range(3):
    if i == 1:
        pass
    print(i)
```

---

````python
choice = input("Enter 'y' or 'n'")

if choice =="y":
    print("You selected yes")
elif choice =="n":
    pass
else:
    print("invalid details")
    

````
