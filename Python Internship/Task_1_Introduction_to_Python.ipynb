# 🐍 Task 1: Introduction to Python

## What is Python?

Python is a **high-level, interpreted, object-oriented, and general-purpose programming language** created by **Guido van Rossum** and first released in **1991**.

It is simple, readable, and widely used for web development, automation, data science, artificial intelligence, machine learning, scripting, and more.

---

## Features of Python

- Easy to learn and read
- Interpreted language
- Open source and free
- Cross-platform
- Object-Oriented Programming support
- Large standard library
- Huge community support
- Dynamically typed

---

## Why is Python Popular?

- Beginner-friendly syntax
- Less code, higher productivity
- Powerful libraries
- Used in AI, ML, Web, Automation, Data Science
- Strong community support

---

## Why Python is Used in AI

- Simple syntax
- Rich AI libraries (NumPy, Pandas, TensorFlow, PyTorch, Scikit-learn)
- Rapid prototyping
- Excellent community support

---

## Python 2 vs Python 3

| Feature | Python 2 | Python 3 |
|---------|----------|----------|
| Release | 2000 | 2008 |
| Status | Deprecated | Active |
| Print | `print "Hi"` | `print("Hi")` |
| Division | `5/2 = 2` | `5/2 = 2.5` |
| Unicode | Limited | Default |

## Short Notes

- Python is high-level and interpreted.
