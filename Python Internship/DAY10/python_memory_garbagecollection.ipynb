# 1. Memory Management

Python automatically manages memory.

It mainly uses:

- **Stack Memory**
- **Heap Memory**

---

# 2. Stack Memory

Stores:

- Function calls
- Local variable references
- Execution frames

Example:

```python
def greet():
    name = "Mayur"
    print(name)

greet()
```

---

# 3. Heap Memory

Stores actual Python objects such as:

- int
- float
- string
- list
- tuple
- dict
- set
- class objects

Example:

```python
numbers = [10, 20, 30]
```

---

# 4. Variables Store References

Variables do **not** store actual values.

They store references to objects.

```python
a = [1, 2, 3]
b = a

print(a)
print(b)

print(id(a))
print(id(b))
```

---

# 5. Object Identity (`id()`)

Every object has:

- Identity
- Type
- Value

```python
x = 100

print(id(x))
print(type(x))
print(x)
```

---

# 6. Equality (`==`) vs Identity (`is`)

`==` → Compare values

`is` → Compare object identity

```python
a = [1, 2]
b = [1, 2]
c = a

print(a == b)
print(a is b)

print(a == c)
print(a is c)
```

---

# 7. Mutable Objects

Objects can be modified without creating a new object.

Examples:

- list
- dict
- set

```python
numbers = [1, 2]

print(id(numbers))

numbers.append(3)

print(numbers)
print(id(numbers))
```

---

# 8. Immutable Objects

Cannot be modified.

Python creates a new object.

Examples:

- int
- float
- bool
- string
- tuple

```python
x = 10

print(id(x))

x = 20

print(id(x))
```

---

# 9. Reference Counting

Python keeps track of how many references point to an object.

```python
import sys

a = []

print(sys.getrefcount(a))

b = a

print(sys.getrefcount(a))

del b

print(sys.getrefcount(a))
```

---

# 10. Small Integer Caching

Python usually reuses integers from **-5 to 256**.

```python
a = 100
b = 100

print(a is b)
```

---

# 11. String Interning

Python often reuses identical strings.

```python
a = "Python"
b = "Python"

print(a is b)
```

---

# 12. Garbage Collection

Garbage Collection removes unreachable objects.

```python
import gc

print(gc.isenabled())
```

---

# 13. Circular References

Reference counting alone cannot remove circular references.

```python
import gc

a = []
b = []

a.append(b)
b.append(a)

del a
del b

print(gc.collect())
```

---

# 14. `gc` Module

### Enable GC

```python
import gc

gc.enable()

print(gc.isenabled())
```

### Disable GC

```python
gc.disable()

print(gc.isenabled())
```

### Manual Garbage Collection

```python
gc.enable()

print(gc.collect())
```

### Current GC Counts

```python
print(gc.get_count())
```

### GC Threshold

```python
print(gc.get_threshold())
```

---

# 15. Weak References

Weak references do not increase reference count.

```python
import weakref

class Student:
    pass

s = Student()

ref = weakref.ref(s)

print(ref())

del s

print(ref())
```

---

# 16. WeakValueDictionary

Values are automatically removed.

```python
import weakref

class Student:
    pass

s = Student()

cache = weakref.WeakValueDictionary()

cache["student"] = s

print(cache)

del s

print(cache)
```

---

# 17. WeakKeyDictionary

Keys are weak references.

```python
import weakref

class Student:
    pass

student = Student()

data = weakref.WeakKeyDictionary()

data[student] = "Python"

print(data)
```

---

# 18. WeakSet

Stores weak references.

```python
import weakref

class Student:
    pass

s = Student()

students = weakref.WeakSet()

students.add(s)

print(list(students))
```

---

# 19. Memory Optimization

### Delete Unused Objects

```python
numbers = list(range(100000))

del numbers
```

### Use Generators

```python
nums = (x for x in range(10))

for i in nums:
    print(i)
```

# 📝 Quick Revision

- Stack → Function calls & local references
- Heap → Actual Python objects
- Variables store references
- `id()` → Object identity
- `==` → Compare values
- `is` → Compare object identity
- Mutable objects → Same object after modification
- Immutable objects → New object after reassignment
- Reference Counting → Object removed when count = 0
- Garbage Collection → Removes circular references
- `gc.collect()` → Manual garbage collection
- `gc.get_count()` → Current GC counts
- `gc.get_threshold()` → GC thresholds
- Weak References → Don't keep objects alive
- WeakValueDictionary → Weak values
- WeakKeyDictionary → Weak keys
- WeakSet → Weak references in a set
- Generators → Memory efficient

# 💻 All Code (Single Code Block)

```python
import sys
import gc
import weakref

# Stack Example
def greet():
    name = "Mayur"
    print(name)

greet()

# Variables Store References
a = [1, 2, 3]
b = a
print(a, b)
print(id(a), id(b))

# Object Identity
x = 100
print(id(x), type(x), x)

# == vs is
a = [1, 2]
b = [1, 2]
c = a
print(a == b)
print(a is b)
print(a == c)
print(a is c)

# Mutable Object
numbers = [1, 2]
print(id(numbers))
numbers.append(3)
print(numbers)
print(id(numbers))

# Immutable Object
x = 10
print(id(x))
x = 20
print(id(x))

# Reference Counting
a = []
print(sys.getrefcount(a))
b = a
print(sys.getrefcount(a))
del b
print(sys.getrefcount(a))

# Small Integer Caching
a = 100
b = 100
print(a is b)

# String Interning
a = "Python"
b = "Python"
print(a is b)

# Garbage Collection
print(gc.isenabled())

# Circular Reference
a = []
b = []
a.append(b)
b.append(a)
del a
del b
print(gc.collect())

# GC Module
gc.enable()
print(gc.isenabled())
gc.disable()
print(gc.isenabled())
gc.enable()
print(gc.collect())
print(gc.get_count())
print(gc.get_threshold())

# Weak Reference
class Student:
    pass

s = Student()
ref = weakref.ref(s)
print(ref())
del s
print(ref())

# WeakValueDictionary
s = Student()
cache = weakref.WeakValueDictionary()
cache["student"] = s
print(cache)
del s
print(cache)

# WeakKeyDictionary
student = Student()
data = weakref.WeakKeyDictionary()
data[student] = "Python"
print(data)

# WeakSet
s = Student()
students = weakref.WeakSet()
students.add(s)
print(list(students))

# Memory Optimization
numbers = list(range(100000))
del numbers

nums = (x for x in range(10))
for i in nums:
    print(i)
```
