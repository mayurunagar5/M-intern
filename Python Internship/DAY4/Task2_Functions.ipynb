# 1. Function Fundamentals

Functions are reusable blocks of code.

``` python
def greet():
    print("Hello")
```

Benefits: - Code reuse - Modularity - Readability - Easier testing

------------------------------------------------------------------------

# 2. Parameters vs Arguments

**Parameters** are variables in the function definition.

``` python
def add(a, b):
    return a + b
```

**Arguments** are actual values passed during the call.

``` python
add(10, 20)
```

------------------------------------------------------------------------

# 3. Parameters

## Positional

``` python
divide(20, 5)
```

## Keyword

``` python
divide(b=5, a=20)
```

## Default

``` python
def greet(name="Guest"):
    ...
```

### Mutable Default Pitfall

Bad:

``` python
def add_item(item, lst=[]):
    lst.append(item)
```

Good:

``` python
def add_item(item, lst=None):
    if lst is None:
        lst = []
```

## Positional-only

``` python
def func(a, /):
    ...
```

## Keyword-only

``` python
def func(*, timeout=10):
    ...
```

------------------------------------------------------------------------

# 4. Return Values

Prefer returning data.

``` python
def square(x):
    return x*x
```

------------------------------------------------------------------------

# 5. \*args & \*\*kwargs

``` python
def func(*args):
    ...
```

Collects positional arguments into a tuple.

``` python
def func(**kwargs):
    ...
```

Collects keyword arguments into a dictionary.

### Packing

``` python
def func(*args, **kwargs):
    ...
```

### Unpacking

``` python
func(*values)
func(**mapping)
```

------------------------------------------------------------------------

# 6. Recursion

Every recursive function has:

-   Base case
-   Recursive case

``` python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)
```

Topics covered: - Call stack - Stack frames - Memoization
(`lru_cache`) - Tail recursion - Recursion vs iteration

------------------------------------------------------------------------

# 7. Lambda Functions

Syntax:

``` python
lambda parameters: expression
```

Example:

``` python
square = lambda x: x*x
```

Common use:

``` python
sorted(students, key=lambda s: s[1])
```

Limitations: - One expression only - No statements - No loops - No
assignments

------------------------------------------------------------------------

# 8. Function Annotations

``` python
def add(a: int, b: int) -> int:
    return a+b
```

Collection types:

``` python
list[int]
dict[str, int]
tuple[int, int]
```

Useful typing constructs: - Any - Optional - Callable - int \| float

Type hints improve documentation and tooling but are **not** enforced at
runtime.

------------------------------------------------------------------------

# 9. Docstrings

``` python
def add(a, b):
    """Return the sum of two numbers."""
```

Access:

``` python
help(add)
add.__doc__
```

Preferred style: Google Style.

------------------------------------------------------------------------

# Best Practices

-   Prefer clear function names.
-   Avoid mutable default arguments.
-   Use keyword arguments for optional settings.
-   Keep lambdas short.
-   Use recursion only when appropriate.
-   Add type hints to reusable code.
-   Write docstrings for public APIs.

------------------------------------------------------------------------

# Common Pitfalls

-   Parameters vs arguments confusion
-   Mutable default arguments
-   Missing recursion base case
-   Assuming Python optimizes tail recursion
-   Overusing lambda
-   Assuming type hints enforce types
-   Confusing comments with docstrings

````python
# recursion 

def count(number):
    if number==1:
        return

    print(number)

    count(number-1)
count(5)
````

````python
def fact(number):
    if number==1:
        return 1

    return number * fact(number -1)
print(fact(5))
````

````python

def sum(number):
    if number== 1:
        return 1

    return number + sum(number - 1)

print(sum(7))
````

````python
def reverse(text):

    if text =="":
        return ""
    return reverse(text[1:]) +text[0]

print(reverse("python"))
````

````python
def count_characters(text):

    if text == "":          # Base Case
        return 0

    return  1+count_characters(text[1:])

print(count_characters("Python"))
````

````python
def maximum(numbers):

    if len(numbers) == 1:   # Base Case
        return numbers[0]

    return max(numbers[0], maximum(numbers[1:]))

print(maximum([12, 45, 7, 98, 33]))
````
